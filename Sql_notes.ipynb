# 6. Constraints

Constraints are rules applied to table columns.

### PRIMARY KEY

Uniquely identifies each row.

```sql
employee_id INT PRIMARY KEY
```

### NOT NULL

Value cannot be NULL.

```sql
name VARCHAR(100) NOT NULL
```

### UNIQUE

Values must be unique.

```sql
email VARCHAR(100) UNIQUE
```

### DEFAULT

Provides a default value.

```sql
status VARCHAR(20) DEFAULT 'Active'
```

### CHECK

Applies a condition.

```sql
age INT CHECK(age >= 18)
```

### FOREIGN KEY

Creates a relationship between tables.

```sql
FOREIGN KEY (department_id)
REFERENCES departments(department_id)
```

# 26. SQL Query Execution Order

Very important for interviews.

```text
FROM
JOIN
WHERE
GROUP BY
HAVING
SELECT
DISTINCT
ORDER BY
LIMIT
```

# 34. INNER JOIN

Returns matching records from both tables.

```sql
SELECT
    e.name,
    d.department_name
FROM employees e
INNER JOIN departments d
ON e.department_id = d.department_id;
```

---

# 35. LEFT JOIN

Returns all records from the left table and matching records from the right table.

```sql
SELECT
    e.name,
    d.department_name
FROM employees e
LEFT JOIN departments d
ON e.department_id = d.department_id;
```

---

# 36. RIGHT JOIN

Returns all records from the right table.

```sql
SELECT
    e.name,
    d.department_name
FROM employees e
RIGHT JOIN departments d
ON e.department_id = d.department_id;
```

---

# 37. FULL OUTER JOIN

Returns matching and non-matching records from both tables.

```sql
SELECT *
FROM employees e
FULL OUTER JOIN departments d
ON e.department_id = d.department_id;
```

---

# 38. SELF JOIN

A table joins with itself.

Example: employee-manager relationship.

```sql
SELECT
    e.name AS employee,
    m.name AS manager
FROM employees e
JOIN employees m
ON e.manager_id = m.employee_id;
```

# 42. Subquery

A query inside another query.

Example:

```sql
SELECT *
FROM employees
WHERE salary >
(
    SELECT AVG(salary)
    FROM employees
);
```

---

# 43. Correlated Subquery

A subquery that depends on the outer query.

```sql
SELECT e1.name, e1.salary
FROM employees e1
WHERE e1.salary >
(
    SELECT AVG(e2.salary)
    FROM employees e2
    WHERE e2.department_id = e1.department_id
);
```

---

# 44. EXISTS

Checks whether a subquery returns records.

```sql
SELECT *
FROM employees e
WHERE EXISTS
(
    SELECT 1
    FROM departments d
    WHERE d.department_id = e.department_id
);
```

# 45. CTE

CTE = Common Table Expression.

```sql
WITH high_salary AS
(
    SELECT *
    FROM employees
    WHERE salary > 70000
)
SELECT *
FROM high_salary;
```

Multiple CTEs:

```sql
WITH
sales_data AS
(
    SELECT *
    FROM sales
),
high_sales AS
(
    SELECT *
    FROM sales_data
    WHERE amount > 10000
)
SELECT *
FROM high_sales;
```

# 46. Window Functions

Window functions perform calculations across related rows **without collapsing rows**.

Basic syntax:

```sql
function() OVER (
    PARTITION BY column
    ORDER BY column
)
```

---

# 47. ROW_NUMBER

Assigns unique row numbers.

```sql
SELECT
    name,
    salary,
    ROW_NUMBER() OVER (
        ORDER BY salary DESC
    ) AS row_num
FROM employees;
```

---

# 48. RANK

Same values receive the same rank, with gaps.

```sql
SELECT
    name,
    salary,
    RANK() OVER (
        ORDER BY salary DESC
    ) AS salary_rank
FROM employees;
```

Example:

```text
Salary    Rank
100000     1
90000      2
90000      2
80000      4
```

---

# 49. DENSE_RANK

Same values receive the same rank, without gaps.

```sql
SELECT
    name,
    salary,
    DENSE_RANK() OVER (
        ORDER BY salary DESC
    ) AS salary_rank
FROM employees;
```

Example:

```text
Salary    Rank
100000     1
90000      2
90000      2
80000      3
```

---

# 50. PARTITION BY

Divides data into groups for window calculations.

```sql
SELECT
    name,
    department,
    salary,
    RANK() OVER (
        PARTITION BY department
        ORDER BY salary DESC
    ) AS rank_in_department
FROM employees;
```

---

# 51. LAG

Gets value from the previous row.

```sql
SELECT
    order_date,
    sales,
    LAG(sales) OVER (
        ORDER BY order_date
    ) AS previous_sales
FROM sales;
```

# 52. LEAD

```sql
SELECT
    order_date,
    sales,
    SUM(sales) OVER (
        ORDER BY order_date
    ) AS running_total
FROM sales;
```

# 62. Important SQL Query Patterns

### Second Highest Salary

```sql
SELECT MAX(salary)
FROM employees
WHERE salary <
(
    SELECT MAX(salary)
    FROM employees
);
```

### Top 3 Salaries

```sql
SELECT DISTINCT salary
FROM employees
ORDER BY salary DESC
LIMIT 3;
```

### Employees Above Average Salary

```sql
SELECT *
FROM employees
WHERE salary >
(
    SELECT AVG(salary)
    FROM employees
);
```

### Duplicate Values

```sql
SELECT name, COUNT(*)
FROM employees
GROUP BY name
HAVING COUNT(*) > 1;
```

### Highest Salary by Department

```sql
SELECT
    department,
    MAX(salary) AS max_salary
FROM employees
GROUP BY department;
```

### Top Employee in Each Department

```sql
WITH ranked_employees AS
(
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY department_id
            ORDER BY salary DESC
        ) AS rn
    FROM employees
)
SELECT *
FROM ranked_employees
WHERE rn = 1;
```
