# ACID PROPERTIES IN SQL

ACID properties are a set of **four principles** that ensure database transactions are **reliable, consistent, and safe**.

```text
A → Atomicity
C → Consistency
I → Isolation
D → Durability
```

---

# 1. Atomicity

**Atomicity means: ALL or NOTHING.**

A transaction is treated as a single unit.

* Either the entire transaction is completed.
* Or none of it is completed.
* If any operation fails, the whole transaction is rolled back.

### Example

Suppose ₹5,000 is transferred from Account A to Account B.

```text
Account A → -₹5,000
Account B → +₹5,000
```

Both operations must succeed.

If money is deducted from A but not added to B:

```text
❌ Transaction fails
↓
ROLLBACK
↓
Money is restored to Account A
```

### SQL Example

```sql
START TRANSACTION;

UPDATE accounts
SET balance = balance - 5000
WHERE account_id = 1;

UPDATE accounts
SET balance = balance + 5000
WHERE account_id = 2;

COMMIT;
```

If something goes wrong:

```sql
ROLLBACK;
```

### Remember

```text
Atomicity = ALL or NOTHING
```

---

# 2. Consistency

**Consistency means that a transaction must take the database from one valid state to another valid state.**

All database rules and constraints must remain satisfied.

### Example

Suppose:

```text
Account Balance = ₹10,000
```

A transaction tries to withdraw:

```text
₹15,000
```

If the database has a rule:

```text
balance >= 0
```

The transaction should fail.

```text
Invalid State ❌
     ↓
Transaction rejected
     ↓
Valid State ✅
```

### Consistency is maintained using:

```text
PRIMARY KEY
FOREIGN KEY
UNIQUE
NOT NULL
CHECK
DEFAULT
```

Example:

```sql
CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    age INT CHECK(age >= 18)
);
```

This prevents invalid data such as:

```text
age = 15 ❌
```

### Remember

```text
Consistency = Database rules are always maintained
```

---

# 3. Isolation

**Isolation means that multiple transactions running at the same time should not incorrectly interfere with each other.**

Each transaction should behave as if it is running independently.

### Example

Two customers try to purchase the **last available product** at the same time.

```text
Customer A → Buys last product
Customer B → Also tries to buy last product
```

Without proper isolation:

```text
Both transactions may think:
"Product is available" ❌
```

With proper isolation:

```text
Transaction A
     ↓
Locks/updates data
     ↓
Transaction B waits
     ↓
A completes
     ↓
B sees updated data
```

### SQL Example

```sql
START TRANSACTION;

SELECT stock
FROM products
WHERE product_id = 101
FOR UPDATE;

UPDATE products
SET stock = stock - 1
WHERE product_id = 101;

COMMIT;
```

### Isolation Levels

```text
READ UNCOMMITTED
        ↓
READ COMMITTED
        ↓
REPEATABLE READ
        ↓
SERIALIZABLE
```

From generally **lower → higher isolation**.

### Remember

```text
Isolation = Concurrent transactions should not interfere incorrectly
```

---

# 4. Durability

**Durability means that once a transaction is committed, the changes are permanently saved.**

Even if the database crashes immediately afterward, committed data should not be lost.

### Example

Suppose:

```sql
UPDATE accounts
SET balance = balance + 5000
WHERE account_id = 1;

COMMIT;
```

After `COMMIT`:

```text
Transaction successful ✅
        ↓
Data permanently saved
        ↓
Database crashes 💥
        ↓
Database restarts
        ↓
₹5,000 change still exists ✅
```

### Remember

```text
Durability = COMMIT means permanent
```

# 8. ACID vs Transaction

### Transaction

A **transaction** is a sequence of database operations treated as one logical unit.

Example:

```sql
START TRANSACTION;

UPDATE accounts
SET balance = balance - 5000
WHERE account_id = 1;

UPDATE accounts
SET balance = balance + 5000
WHERE account_id = 2;

COMMIT;
```

### ACID

ACID defines the properties that make the transaction reliable.

```text
Transaction
     ↓
┌───────────────────────────┐
│ A → Atomicity             │
│ C → Consistency            │
│ I → Isolation              │
│ D → Durability             │
└───────────────────────────┘
```

# 9. Interview Answer

### What are ACID properties?

**ACID stands for Atomicity, Consistency, Isolation, and Durability.**

* **Atomicity** → A transaction is completed completely or not at all.
* **Consistency** → A transaction keeps the database in a valid state.
* **Isolation** → Concurrent transactions should not interfere incorrectly.
* **Durability** → Once a transaction is committed, its changes are permanently stored.

### One-line answer:

```text
ACID properties ensure that database transactions are reliable,
consistent, isolated, and permanent.
```
