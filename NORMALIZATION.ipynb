# DATABASE NORMALIZATION

Normalization is the process of **organizing data in a database** to:

* Reduce data redundancy (duplicate data)
* Improve data consistency
* Eliminate data anomalies
* Make the database easier to maintain

### Simple Definition

```text id="a1b2c3"
Normalization is the process of organizing database tables
to reduce redundancy and improve data integrity.
```

---

# Why Do We Need Normalization?

Without normalization:

* Duplicate data is stored.
* More storage is used.
* Updating data becomes difficult.
* Data inconsistencies occur.
* Insert, update, and delete anomalies can occur.

### Example (Unnormalized Table)

| Student_ID | Student_Name | Course | Instructor |
| ---------- | ------------ | ------ | ---------- |
| 101        | Rahul        | SQL    | Amit       |
| 101        | Rahul        | Python | Neha       |
| 102        | Priya        | SQL    | Amit       |

Problems:

* Rahul's name is repeated.
* Amit's name is repeated.
* Updating Rahul's name requires changing multiple rows.

---

# What is Redundancy?

Redundancy means storing the **same information multiple times**.

Example:

```text id="d4e5f6"
Student Name = Rahul

101 Rahul SQL
101 Rahul Python
101 Rahul Excel
```

Here, "Rahul" is repeated in every row.

---

# Data Anomalies

Normalization removes three types of anomalies.

---

# 1. Insert Anomaly

Unable to insert data without unnecessary information.

### Example

Suppose no student has enrolled in the AI course yet.

You cannot insert:

```text id="g7h8i9"
Course = AI
Instructor = John
```

because Student_ID is required.

---

# 2. Update Anomaly

Updating the same information in multiple rows.

Example:

```text id="j1k2l3"
Instructor = Amit
```

appears in 100 rows.

If Amit changes to Raj:

```text id="m4n5o6"
Need to update all 100 rows.
```

Missing one row creates inconsistent data.

---

# 3. Delete Anomaly

Deleting one row accidentally removes useful information.

Example:

If Rahul is the only student enrolled in SQL:

```text id="p7q8r9"
Delete Rahul
        ↓
SQL course information is also lost.
```

---

# Goals of Normalization

```text id="s1t2u3"
Reduce Redundancy
Improve Consistency
Remove Anomalies

Save Storage
```

---

# Normal Forms

```text id="v4w5x6"
UNF
 ↓
1NF
 ↓
2NF
 ↓
3NF
 ↓
BCNF
 ↓
4NF
 ↓
5NF
```

Most real-world databases use **3NF or BCNF**.

---

# UNF (Unnormalized Form)

Data contains repeating groups or multiple values in one column.

Example:

| Student_ID | Name  | Courses            |
| ---------- | ----- | ------------------ |
| 101        | Rahul | SQL, Python, Excel |

Problem:

```text id="y7z8a9"
One column contains multiple values.
```

---

# First Normal Form (1NF)

## Rule

Every column must contain **atomic (single) values**.

There should be:

* No repeating groups
* No multiple values in one cell

### Before 1NF

| Student | Courses     |
| ------- | ----------- |
| Rahul   | SQL, Python |

### After 1NF

| Student | Course |
| ------- | ------ |
| Rahul   | SQL    |
| Rahul   | Python |

### Rules

```text id="b1c2d3"
Each cell contains one value.
No repeating columns.
Each row is unique.
```

---

# Second Normal Form (2NF)

## Rule

Table must already be in **1NF**.

Remove **Partial Dependency**.

---

# What is Partial Dependency?

A non-key column depends on **only part of a composite primary key**.

Suppose:

Primary Key:

```text id="e4f5g6"
(Student_ID, Course_ID)
```

Table:

| Student_ID | Course_ID | Student_Name | Course_Name |
| ---------- | --------- | ------------ | ----------- |

Problem:

```text id="h7i8j9"
Student_Name depends only on Student_ID.
Course_Name depends only on Course_ID.
```

Not on the entire key.

---

# Solution

Split into multiple tables.

### Students

| Student_ID | Student_Name |
| ---------- | ------------ |
| 101        | Rahul        |

### Courses

| Course_ID | Course_Name |
| --------- | ----------- |
| 201       | SQL         |

### Enrollment

| Student_ID | Course_ID |
| ---------- | --------- |

Now every non-key column depends on the whole key.

### Remember

```text id="k1l2m3"
2NF = Remove Partial Dependency
```

---

# Third Normal Form (3NF)

## Rule

Table must already be in **2NF**.

Remove **Transitive Dependency**.

---

# What is Transitive Dependency?

A non-key column depends on another non-key column instead of the primary key.

Example:

| Emp_ID | Emp_Name | Dept_ID | Dept_Name |
| ------ | -------- | ------- | --------- |

Primary Key:

```text id="n4o5p6"
Emp_ID
```

Dependency:

```text id="q7r8s9"
Emp_ID
   ↓
Dept_ID
   ↓
Dept_Name
```

Dept_Name depends on Dept_ID, not directly on Emp_ID.

---

# Solution

### Employees

| Emp_ID | Emp_Name | Dept_ID |
| ------ | -------- | ------- |

### Departments

| Dept_ID | Dept_Name |
| ------- | --------- |

Now:

```text id="t1u2v3"
Dept_Name depends only on Dept_ID.
```

### Remember

```text id="w4x5y6"
3NF = Remove Transitive Dependency
```

---

# Boyce-Codd Normal Form (BCNF)

BCNF is a stronger version of 3NF.

## Rule

Every determinant must be a candidate key.

Simple definition:

```text id="z7a8b9"
Every dependency should be based on a candidate key.
```

BCNF removes some anomalies that 3NF cannot.

---

# Fourth Normal Form (4NF)

Removes **Multi-Valued Dependencies**.

Example:

| Student | Hobby | Language |
| ------- | ----- | -------- |

Rahul has:

```text id="c1d2e3"
Hobby:
Cricket
Music

Languages:
English
Hindi
```

All combinations are stored unnecessarily.

Solution:

Separate tables.

### Student_Hobby

| Student | Hobby |

### Student_Language

| Student | Language |

---

# Fifth Normal Form (5NF)

Removes **Join Dependencies**.

Complex tables are divided into smaller tables so that joining them reconstructs the original data without redundancy.

Mostly used in advanced database design.

---

# Summary of Normal Forms

| Normal Form | Removes                              |
| ----------- | ------------------------------------ |
| 1NF         | Repeating groups and multiple values |
| 2NF         | Partial dependency                   |
| 3NF         | Transitive dependency                |
| BCNF        | Non-candidate key dependency         |
| 4NF         | Multi-valued dependency              |
| 5NF         | Join dependency                      |

---

# Dependency Flow

```text id="f4g5h6"
1NF
 ↓
Atomic Values

2NF
 ↓
No Partial Dependency

3NF
 ↓
No Transitive Dependency

BCNF
 ↓
Every Determinant is a Candidate Key

4NF
 ↓
No Multi-Valued Dependency

5NF
 ↓
No Join Dependency
```

# Advantages of Normalization

```text id="i7j8k9"
Less Redundancy
Better Consistency
Less Storage
Easy Maintenance
Improved Data Integrity
Fewer Anomalies
```

---

# Disadvantages of Normalization

```text id="l1m2n3"
More Tables
More JOIN Operations
Slightly Slower Read Queries
More Complex Database Design
```

---

# Normalization vs Denormalization

| Normalization         | Denormalization              |
| --------------------- | ---------------------------- |
| Reduces redundancy    | Increases redundancy         |
| More tables           | Fewer tables                 |
| More JOINs            | Fewer JOINs                  |
| Better data integrity | Faster read performance      |
| Preferred for OLTP    | Preferred for OLAP/Reporting |

# Real-Life Example

Suppose a company stores employee and department information.

### Before Normalization

| Emp_ID | Name  | Dept_ID | Dept_Name |
| ------ | ----- | ------- | --------- |
| 101    | Rahul | 10      | IT        |
| 102    | Priya | 10      | IT        |
| 103    | Amit  | 20      | HR        |

Problem:

```text id="o4p5q6"
Department names are repeated.
```

### After Normalization

Employees Table

| Emp_ID | Name | Dept_ID |
| ------ | ---- | ------- |

Departments Table

| Dept_ID | Dept_Name |
| ------- | --------- |

Benefits:

```text id="r7s8t9"
Department information stored only once.
Easy to update.
No redundancy.
```

### Difference Between 2NF and 3NF

| 2NF                        | 3NF                           |
| -------------------------- | ----------------------------- |
| Removes partial dependency | Removes transitive dependency |
| Requires 1NF               | Requires 2NF                  |

### Difference Between 3NF and BCNF

| 3NF                       | BCNF                   |
| ------------------------- | ---------------------- |
| Less strict               | More strict            |
| Some anomalies may remain | Removes more anomalies |
